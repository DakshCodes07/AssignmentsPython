```python
def is_right_angled(a, b, c):
    sides = sorted([a, b, c])
    x, y, z = sides
    return round(x**2 + y**2, 5) == round(z**2, 5)

a = float(input("Enter side a: "))
b = float(input("Enter side b: "))
c = float(input("Enter side c: "))

if is_right_angled(a, b, c):
    print("The triangle is right-angled.")
else:
    print("The triangle is NOT right-angled.")
```

```python
def prime_factors(n):
    factors = []
    d = 2
    while d * d <= n:
        while n % d == 0:
            factors.append(d)
            n //= d
        d += 1
    if n > 1:
        factors.append(n)
    return sorted(set(factors))

num = int(input("Input number: "))
print("prime factors =", prime_factors(num))
```

```python
import math

def quadratic_roots(a, b, c):
    try:
        d = b**2 - 4 * a * c
        root1 = (-b + math.sqrt(d)) / (2 * a)
        root2 = (-b - math.sqrt(d)) / (2 * a)
        print("Roots are:", root1, root2)
    except ValueError:
        print("Error: math domain error (discriminant is negative — complex roots).")
    except NameError:
        print("Error: a variable used is not defined.")
    except TypeError:
        print("Error: invalid input type — coefficients must be numbers.")
    except ZeroDivisionError:
        print("Error: coefficient 'a' cannot be zero.")

a = float(input("Enter coefficient a: "))
b = float(input("Enter coefficient b: "))
c = float(input("Enter coefficient c: "))
quadratic_roots(a, b, c)
```
