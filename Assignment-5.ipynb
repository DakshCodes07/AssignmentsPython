```python
# ---------- Basics ----------
s = "Hello"
print(s[1])  # e

s1 = s + s[0]
print(s1)  # HelloH

# ---------- Loop through string ----------
for x in "banana":
    print(x)

# ---------- len() ----------
a = "Hello, World!"
print(len(a))  # 13

# ---------- Check substring ----------
txt = "The best things in life are free!"
print("free" in txt)  # True

# ---------- Concatenation ----------
a = "Hello"
b = "World!"
c = a + b
print(c)  # HelloWorld!

# ---------- String methods ----------
a = "Hello, World!"
print(a.upper())  # HELLO, WORLD!

a = " Hello, World! "
print(a.strip())  # Hello, World!

a = "Hello, World!"
print(a.replace("H", "J"))  # Jello, World!

a = "Hello, World!"
print(a.split(","))  # ['Hello', ' World!']

# ---------- Slicing ----------
s = "GoodMorning"
print(s[1:4])   # ood
print(s[:3])    # Goo
print(s[3:])    # dMorning

# ---------- Immutability ----------
s = "goodMorning"
s = "G" + s[1:]
print(s)  # GoodMorning

# ---------- Deleting a string ----------
s = "GfG"
del s

# ---------- Updating a string ----------
s = "hello mit"
s1 = "H" + s[1:]
s2 = s.replace("mit", "MIT WPU")
print(s1)  # Hello mit
print(s2)  # hello MIT WPU

# ---------- Regex: starts/ends with ----------
import re

txt = "The rain in Spain"
x = re.search("^The.*Spain$", txt)
if x:
    print("YES! We have a match!")
else:
    print("No match")

# ---------- Regex: check allowed characters ----------
def check_string(s):
    pattern = "^[A-Za-z0-9]+$"  # ^ = start, $ = end, + = one or more
    return bool(re.match(pattern, s))

s1 = "Hello123"
s2 = "Hello_123"  # contains underscore
print(check_string(s1))  # True
print(check_string(s2))  # False

# ---------- Alternative: isalnum() ----------
def check_string_alnum(s):
    return s.isalnum()

s1 = "Hello123"
s2 = "Hello 123!"  # contains space and special character
print(check_string_alnum(s1))  # True
print(check_string_alnum(s2))  # False
```
